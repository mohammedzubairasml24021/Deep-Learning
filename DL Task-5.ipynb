```python
import numpy as np
np.set_printoptions(precision=3, suppress=True)

# example: 4 samples (rows), 3 features (columns)
X = np.array([
    [1.0, 2.0, 30.0],
    [2.0, 4.0, 32.0],
    [3.0, 6.0, 28.0],
    [4.0, 8.0, 34.0],
])

print("Input data:\n", X)
```

```python
eps = 1e-5

mean = X.mean(axis=0)    # mean of each column (feature), across the 4 rows
var = X.var(axis=0)      # variance of each column

print("mean per feature:", mean)
print("var per feature :", var)

x_hat = (X - mean) / np.sqrt(var + eps)

print("\nNormalized X (x_hat):\n", x_hat)
print("\nnew mean per feature (should be ~0):", x_hat.mean(axis=0))
print("new var per feature (should be ~1):", x_hat.var(axis=0))
```

```python
gamma = np.ones(3)     # start at 1 (no stretch)
beta = np.zeros(3)     # start at 0 (no shift)

out = gamma * x_hat + beta

print("Output after gamma/beta:\n", out)
```

```python
mean_ln = X.mean(axis=1, keepdims=True)    # mean of each ROW (sample), across its features
var_ln = X.var(axis=1, keepdims=True)

x_hat_ln = (X - mean_ln) / np.sqrt(var_ln + eps)

print("LayerNorm normalized X:\n", x_hat_ln)
print("\nrow means (should be ~0):", x_hat_ln.mean(axis=1))
print("row vars (should be ~1) :", x_hat_ln.var(axis=1))
```
